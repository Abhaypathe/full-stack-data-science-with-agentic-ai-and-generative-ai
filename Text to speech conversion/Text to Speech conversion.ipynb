```python
import pyttsx3
from IPython.display import Audio, display

text = '''Hello, my name is Abhay. Welcome to the Text-to-Speech conversion system.

Text-to-Speech technology converts written text into natural-sounding spoken audio. It is widely used in virtual assistants, accessibility tools, navigation systems, e-learning applications, and customer service.

Today is a beautiful day to learn something new. Artificial intelligence is making communication between humans and computers easier, faster, and more natural.

Thank you for listening. Have a great day!'''

audio = pyttsx3.init()

audio.setProperty('rate', 150)
audio.setProperty('volume', 1.0)

# Get available voices
voices = audio.getProperty('voices')

# Male voice
audio.setProperty('voice', voices[0].id)

# Convert text to speech
audio.say(text)

# Save audio
audio.save_to_file(text, 'male_voice.mp3')

audio.runAndWait()

# Play audio in Jupyter Notebook
display(Audio('male_voice.mp3'))
```
