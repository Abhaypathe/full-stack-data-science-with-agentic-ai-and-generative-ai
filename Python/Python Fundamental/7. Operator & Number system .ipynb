 # 5.IDENTITY OPERATORS
        -Identity operators are used to check if two values (or variables) are indicating to same object or not.
        - It returns output as boolean
        -is, is not are the identity operator in python
        - is operator (# is- True if the operands are not identical)
        -is not operator (# is not - True if the operands are not identical)    

```python
a=[1,2,3,4]
b=a.copy()
a is b
```

```python
a=5
b=5
print(a is b)
#id (a)== id(b)
```

```python
s1="Satish"
s2="Satish"
print(s1 is s2)
```

```python
a=5
b=3
a is not b
```
