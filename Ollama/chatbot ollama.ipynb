```python
import ollama

models = {
    "1": "deepseek-r1:1.5b",
    "2": "gemma3:270m"
}

print("🤖 Ollama Chatbot")
print()
print("Choose a model:")
print("1. deepseek-r1:1.5b")
print("2. gemma3:270m")

choice = input("\nEnter 1 or 2: ")

if choice not in models:
    print("Invalid choice.")
else:
    model = models[choice]

    print(f"\nUsing: {model}")
    print("Type 'exit' to stop.\n")

    messages = []

    while True:
        user_input = input("You: ")

        if user_input.lower() == "exit":
            print("Goodbye!")
            break

        messages.append({
            "role": "user",
            "content": user_input
        })

        response = ollama.chat(
            model=model,
            messages=messages
        )

        answer = response["message"]["content"]

        print("\nAI:", answer, "\n")

        messages.append({
            "role": "assistant",
            "content": answer
        })
```
