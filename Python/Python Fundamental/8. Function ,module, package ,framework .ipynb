# Function
*A reusable block of code that performs a single specific task.
*A single tool (e.g., a hammer).
*You call it whenever you need it.

# Module
*A single file containing related functions, classes, or variables.
*A small toolbox packed for a specific job.
*You import it and use its contents.

# Package
*A folder containing multiple related modules grouped together.
*A storage container holding multiple toolboxes.
*You install and import it into your code.

# Framework
* A comprehensive skeleton environment that dictates how to build your app.
* A pre-built house structure where you only install the interior.
* The Framework controls the flow and calls your code.
