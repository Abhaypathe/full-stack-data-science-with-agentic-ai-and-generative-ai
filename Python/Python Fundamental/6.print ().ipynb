## print() with format method

```python
num1=20
num2=30

add=num1+num2
print('The addition of {} and {} is= {}'.format(num1,num2,add))
```

## (fstring method)
* variable can bein curly braces
* add write everything inside quotes ''
* at starting simply add f

```python
num1=20
num2=30

add=num1+num2
print(f'The addition of{num1},{num2} is={add}')
```
