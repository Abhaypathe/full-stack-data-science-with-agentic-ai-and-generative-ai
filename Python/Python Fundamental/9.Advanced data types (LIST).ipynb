# list.copy()
used to create  a new  list object with a copy of item

```python
a=[1,2,3,4]
b=a
b.append(5)
print(a)
print(b)          #deep copy
```

```python
#shallow copy
a=[1,2,3,4]
b=a.copy()
b.append(5)
print(a)
print(b)   
```

# deep copy: in the same object 

# shallow copy: values are copied but in new object
